# Adult RHT analysis - RHT axon distribution in the SCN (WT vs *Opn4* KO)

1D spatial-profile analysis of retinohypothalamic tract (RHT) axon signal in the suprachiasmatic nucleus (SCN), wild-type vs melanopsin (*Opn4*) knockout. Single optical sections. Dual-eye CTB tracing; the **red** and **green** channels report the two eyes' projections within the same section.

## Result

No spatial-pattern difference in RHT innervation between WT and KO, in either channel. The spatial-pattern (normalized) profiles are the analysis metric.

## Statistics

Single optical section, outliers removed (see below). Spatial-pattern (normalized) profiles; WT vs KO by Mann-Whitney U with Cohen's d and post-hoc power.

#### Red channel (n = 11 WT vs 7 KO)

| SCN half | p | power | Cohen's d | result |
|---|---|---|---|---|
| negative | 0.5962 | 0.07 | +0.21 | n.s. |
| positive | 0.7914 | 0.06 | +0.12 | n.s. |


#### Green channel (n = 5 WT vs 7 KO)

| SCN half | p | power | Cohen's d | result |
|---|---|---|---|---|
| negative | 0.6389 | 0.05 | +0.09 | n.s. |
| positive | 0.6389 | 0.10 | +0.40 | n.s. |


Red and green both show no spatial-pattern difference between WT and KO (non-significant, small effect sizes).

## Quality control and exclusions

Green sections were assessed by eye and excluded for low signal-to-noise or tissue damage. The exclusions follow the data structure and standard practice in the field.

**Excluded (green):** WT 927, 928, 929, 930, 931, 932; KO 937. **Excluded (red):** KO 937.

The green (CTB-488) channel has low signal-to-noise; many green sections were rejected on quality. The montages below show the section quality directly.
